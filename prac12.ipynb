```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import f_oneway
import statsmodels.api as sm
from statsmodels.formula.api import ols

# LOAD DATASET
file_path = "Retail and wherehouse Sale - Retail and wherehouse Sale.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
-
# PART A – ANALYSIS OF VARIANCE (ANOVA)
# 1. Display first five records
print("\n1. First Five Records:")
print(df.head())

# 2. Number of rows and columns
print("\n2. Dataset Shape:")
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# 3. Column names and data types
print("\n3. Column Names and Data Types:")
print(df.dtypes)

# 4. Missing values and duplicate records
print("\n4. Missing Values:")
print(df.isnull().sum())

print("\nNumber of Duplicate Records:")
print(df.duplicated().sum())

# 5. Different ITEM TYPE categories
print("\n5. Different ITEM TYPE Categories:")
print(df["ITEM TYPE"].unique())

print("\nNumber of ITEM TYPE Categories:")
print(df["ITEM TYPE"].nunique())

# 6. Average RETAIL SALES for each ITEM TYPE
average_sales = df.groupby("ITEM TYPE")["RETAIL SALES"].mean()

print("\n6. Average RETAIL SALES for Each ITEM TYPE:")
print(average_sales)

# 7. Box plot
plt.figure(figsize=(12, 6))
sns.boxplot(
    x="ITEM TYPE",
    y="RETAIL SALES",
    data=df
)

plt.title("Retail Sales by Item Type")
plt.xlabel("Item Type")
plt.ylabel("Retail Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 8. Hypotheses
print("\n8. Hypotheses:")
print("H0: The average RETAIL SALES are equal for all ITEM TYPE categories.")
print("H1: At least one ITEM TYPE has a different average RETAIL SALES.")


# 9 & 10. One-way ANOVA
groups = [
    group["RETAIL SALES"].dropna()
    for name, group in df.groupby("ITEM TYPE")
]

f_statistic, p_value = f_oneway(*groups)

print("\n9 & 10. One-Way ANOVA Results:")
print("F-statistic:", f_statistic)
print("p-value:", p_value)

# 11 & 12. Compare p-value with 0.05
alpha = 0.05

print("\n11 & 12. Hypothesis Test:")

if p_value < alpha:
    print("p-value < 0.05")
    print("Reject the Null Hypothesis (H0).")
else:
    print("p-value >= 0.05")
    print("Do Not Reject the Null Hypothesis (H0).")

# 13. Conclusion
print("\n13. Conclusion:")

if p_value < alpha:
    print(
        "There is a statistically significant difference in average "
        "RETAIL SALES among the different ITEM TYPE categories."
    )
else:
    print(
        "There is no statistically significant difference in average "
        "RETAIL SALES among the different ITEM TYPE categories."
    )
print("Khushi T032")
```
