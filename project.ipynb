```python


# -------------------------
# 1. IMPORT LIBRARIES
# -------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

from scipy import stats
from scipy.optimize import linprog
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.stats.proportion import proportions_ztest

warnings.filterwarnings("ignore")

print("="*70)
print("APPLIED BUSINESS ANALYTICS PROJECT")
print("SUPERSTORE BUSINESS ANALYTICS")
print("="*70)


# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv("Superstore_Business_Analytics_Project.csv")

print("\nDATASET LOADED SUCCESSFULLY")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nFirst 5 Records:")
display(df.head())


# ============================================================
# 3. DATASET INFORMATION
# ============================================================

print("\nDATASET INFORMATION")
print("="*50)

df.info()

print("\nColumn Names:")
print(df.columns.tolist())


# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

print("\nDATA PREPROCESSING")
print("="*50)

# Convert dates
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# Calculate shipping days
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

# Create month column
df["Order Month"] = df["Order Date"].dt.to_period("M").astype(str)

# Create profit margin
df["Profit Margin"] = (
    df["Profit"] / df["Sales"] * 100
)

# Remove duplicate rows
df = df.drop_duplicates()

print("Preprocessing completed.")
print("Rows after preprocessing:", len(df))


# ============================================================
# 5. MISSING VALUES
# ============================================================

print("\nMISSING VALUES")
print("="*50)

missing = df.isnull().sum()
print(missing)


# ============================================================
# 6. DUPLICATE VALUES
# ============================================================

print("\nDUPLICATE RECORDS")
print("="*50)

print("Duplicate rows:", df.duplicated().sum())


# ============================================================
# 7. EXPLORATORY DATA ANALYSIS
# ============================================================

print("\nEXPLORATORY DATA ANALYSIS")
print("="*50)

print("\nBasic Statistics:")
display(df.describe())

print("\nUnique Categories:")
print(df["Category"].unique())

print("\nUnique Regions:")
print(df["Region"].unique())

print("\nUnique Segments:")
print(df["Segment"].unique())

print("\nNumber of Customers:")
print(df["Customer ID"].nunique())

print("\nNumber of Orders:")
print(df["Order ID"].nunique())


# ============================================================
# 8. SALES BY CATEGORY
# ============================================================

category_sales = df.groupby("Category")["Sales"].sum().sort_values(
    ascending=False
)

print("\nSALES BY CATEGORY")
display(category_sales)

plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()


# ============================================================
# 9. PROFIT BY REGION
# ============================================================

region_profit = df.groupby("Region")["Profit"].sum().sort_values(
    ascending=False
)

print("\nPROFIT BY REGION")
display(region_profit)

plt.figure(figsize=(8,5))
region_profit.plot(kind="bar")
plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()


# ============================================================
# 10. DESCRIPTIVE STATISTICS
# ============================================================

print("\nDESCRIPTIVE STATISTICS")
print("="*50)

statistics = df[[
    "Sales",
    "Quantity",
    "Discount",
    "Profit",
    "Shipping Days",
    "Profit Margin"
]].agg([
    "count",
    "mean",
    "median",
    "std",
    "min",
    "max"
])

display(statistics)


# ============================================================
# 11. DISTRIBUTION ANALYSIS
# ============================================================

print("\nDISTRIBUTION ANALYSIS")
print("="*50)

print("Sales Skewness:", df["Sales"].skew())
print("Profit Skewness:", df["Profit"].skew())

plt.figure(figsize=(8,5))
plt.hist(df["Sales"], bins=30)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(8,5))
plt.hist(df["Profit"], bins=30)
plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.show()


# ============================================================
# 12. CORRELATION ANALYSIS
# ============================================================

print("\nCORRELATION ANALYSIS")
print("="*50)

numeric_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit",
    "Shipping Days",
    "Profit Margin"
]

correlation = df[numeric_columns].corr()

display(correlation)

plt.figure(figsize=(9,6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Matrix")
plt.show()


# ============================================================
# 13. REGRESSION ANALYSIS
# ============================================================

print("\nREGRESSION ANALYSIS")
print("="*50)

# Simple regression
X = df["Sales"]
y = df["Profit"]

X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print(model.summary())


# Multiple regression
print("\nMULTIPLE REGRESSION")

X_multi = df[
    ["Sales", "Quantity", "Discount"]
]

X_multi = sm.add_constant(X_multi)

model_multi = sm.OLS(
    df["Profit"],
    X_multi
).fit()

print(model_multi.summary())


# ============================================================
# 14. BUSINESS INTELLIGENCE DASHBOARD DATA
# ============================================================

print("\nBUSINESS INTELLIGENCE DASHBOARD")
print("="*50)

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Quantity:", total_quantity)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)

# Monthly Sales
monthly_sales = df.groupby("Order Month")["Sales"].sum()

plt.figure(figsize=(12,5))
monthly_sales.plot()
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()

# Category
plt.figure(figsize=(7,5))
df.groupby("Category")["Sales"].sum().plot(kind="bar")
plt.title("Sales by Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()

# Region
plt.figure(figsize=(7,5))
df.groupby("Region")["Profit"].sum().plot(kind="bar")
plt.title("Profit by Region")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()


# ============================================================
# 15. PROBABILITY
# ============================================================

print("\nPROBABILITY ANALYSIS")
print("="*50)

# Probability of profit
prob_profit = (df["Profit"] > 0).mean()

# Probability of loss
prob_loss = (df["Profit"] < 0).mean()

# Probability of high discount
prob_high_discount = (df["Discount"] > 0.20).mean()

print("Probability of Profit:", round(prob_profit, 4))
print("Probability of Loss:", round(prob_loss, 4))
print(
    "Probability of Discount > 20%:",
    round(prob_high_discount, 4)
)


# ============================================================
# 16. DECISION MAKING UNDER UNCERTAINTY
# ============================================================

print("\nDECISION MAKING UNDER UNCERTAINTY")
print("="*50)

# Illustrative business decision table
decision_data = pd.DataFrame({
    "Decision": [
        "Increase Marketing",
        "Maintain Marketing",
        "Reduce Marketing"
    ],
    "High_Demand": [50000, 35000, 15000],
    "Medium_Demand": [25000, 30000, 18000],
    "Low_Demand": [-10000, 10000, 12000]
})

print("Illustrative Payoff Table:")
display(decision_data)

probabilities = {
    "High_Demand": 0.30,
    "Medium_Demand": 0.50,
    "Low_Demand": 0.20
}

decision_data["Expected_Value"] = (
    decision_data["High_Demand"] * probabilities["High_Demand"]
    + decision_data["Medium_Demand"] * probabilities["Medium_Demand"]
    + decision_data["Low_Demand"] * probabilities["Low_Demand"]
)

display(decision_data)

print(
    "Expected values are calculated using assumed business probabilities."
)


# ============================================================
# 17. SAMPLING
# ============================================================

print("\nSAMPLING AND SAMPLING DISTRIBUTION")
print("="*50)

np.random.seed(42)

sample_means = []

for i in range(1000):
    sample = df["Sales"].sample(
        n=50,
        replace=True
    )
    sample_means.append(sample.mean())

sample_means = np.array(sample_means)

print("Number of Samples:", len(sample_means))
print("Mean of Sample Means:", sample_means.mean())
print("Standard Deviation:", sample_means.std())

plt.figure(figsize=(8,5))
plt.hist(sample_means, bins=30)
plt.title("Sampling Distribution of Mean Sales")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()


# ============================================================
# 18. CONFIDENCE INTERVAL
# ============================================================

print("\nCONFIDENCE INTERVAL ESTIMATION")
print("="*50)

sales = df["Sales"]

mean_sales = sales.mean()
std_sales = sales.std()
n = len(sales)

for confidence in [0.90, 0.95, 0.99]:

    alpha = 1 - confidence

    t_value = stats.t.ppf(
        1 - alpha/2,
        df=n-1
    )

    margin = t_value * std_sales / np.sqrt(n)

    lower = mean_sales - margin
    upper = mean_sales + margin

    print(
        f"{int(confidence*100)}% CI: "
        f"({lower:.2f}, {upper:.2f})"
    )


# ============================================================
# 19. BOOTSTRAPPING
# ============================================================

print("\nBOOTSTRAPPING")
print("="*50)

np.random.seed(42)

bootstrap_means = []

for i in range(1000):

    sample = np.random.choice(
        df["Sales"],
        size=len(df),
        replace=True
    )

    bootstrap_means.append(
        np.mean(sample)
    )

lower_boot = np.percentile(
    bootstrap_means,
    2.5
)

upper_boot = np.percentile(
    bootstrap_means,
    97.5
)

print("Bootstrap Mean:", np.mean(bootstrap_means))
print(
    "95% Bootstrap CI:",
    (lower_boot, upper_boot)
)

plt.figure(figsize=(8,5))
plt.hist(bootstrap_means, bins=30)
plt.title("Bootstrap Distribution of Mean Sales")
plt.xlabel("Mean Sales")
plt.ylabel("Frequency")
plt.show()


# ============================================================
# 20. HYPOTHESIS TESTING
# ============================================================

print("\nHYPOTHESIS TESTING")
print("="*50)

# H0: Mean profit = 0
# H1: Mean profit != 0

t_stat, p_value = stats.ttest_1samp(
    df["Profit"],
    0
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print(
        "Result: Reject H0. Mean profit is significantly different from zero."
    )
else:
    print(
        "Result: Fail to reject H0."
    )


# ============================================================
# 21. A/B TESTING
# ============================================================

print("\nA/B TESTING")
print("="*50)

# Simulated example because original dataset
# does not contain randomized A/B groups

group_A_visitors = 5000
group_B_visitors = 5000

group_A_conversions = 350
group_B_conversions = 410

conversion_A = (
    group_A_conversions /
    group_A_visitors
)

conversion_B = (
    group_B_conversions /
    group_B_visitors
)

print("Group A Conversion Rate:", conversion_A)
print("Group B Conversion Rate:", conversion_B)

successes = np.array([
    group_A_conversions,
    group_B_conversions
])

observations = np.array([
    group_A_visitors,
    group_B_visitors
])

z_stat, p_value = proportions_ztest(
    successes,
    observations
)

print("Z-statistic:", z_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print(
        "The difference between the two conversion rates is statistically significant."
    )
else:
    print(
        "The difference is not statistically significant."
    )


# ============================================================
# 22. TIME SERIES ANALYSIS
# ============================================================

print("\nTIME SERIES ANALYSIS")
print("="*50)

monthly_sales_df = (
    df.groupby("Order Month")["Sales"]
    .sum()
    .reset_index()
)

monthly_sales_df["Order Month"] = pd.to_datetime(
    monthly_sales_df["Order Month"]
)

monthly_sales_df = monthly_sales_df.set_index(
    "Order Month"
)

monthly_sales_series = monthly_sales_df["Sales"]

plt.figure(figsize=(12,5))
plt.plot(
    monthly_sales_series
)
plt.title("Monthly Sales Time Series")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()


# ============================================================
# 23. EXPONENTIAL SMOOTHING FORECAST
# ============================================================

print("\nEXPONENTIAL SMOOTHING FORECAST")
print("="*50)

model_es = ExponentialSmoothing(
    monthly_sales_series,
    trend="add",
    seasonal=None
).fit()

forecast_es = model_es.forecast(6)

print("Next 6 Month Forecast:")
display(forecast_es)

plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales_series,
    label="Actual"
)

plt.plot(
    forecast_es,
    label="Forecast"
)

plt.title("Exponential Smoothing Forecast")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.show()


# ============================================================
# 24. ARIMA FORECAST
# ============================================================

print("\nARIMA FORECAST")
print("="*50)

arima_model = ARIMA(
    monthly_sales_series,
    order=(1,1,1)
).fit()

arima_forecast = arima_model.forecast(
    steps=6
)

print("ARIMA Forecast:")
display(arima_forecast)

plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales_series,
    label="Actual"
)

plt.plot(
    arima_forecast,
    label="ARIMA Forecast"
)

plt.title("ARIMA Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.show()


# ============================================================
# 25. FORECAST ACCURACY
# ============================================================

print("\nFORECAST ACCURACY")
print("="*50)

# Hold out last 6 months
train = monthly_sales_series.iloc[:-6]
test = monthly_sales_series.iloc[-6:]

accuracy_model = ExponentialSmoothing(
    train,
    trend="add",
    seasonal=None
).fit()

prediction = accuracy_model.forecast(6)

mae = np.mean(
    np.abs(test.values - prediction.values)
)

mse = np.mean(
    (test.values - prediction.values) ** 2
)

rmse = np.sqrt(mse)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)


# ============================================================
# 26. OPTIMIZATION MODELING
# ============================================================

print("\nOPTIMIZATION MODELING")
print("="*50)

# Illustrative product allocation problem
# Maximize profit from three product groups

profit_per_unit = np.array([
    40,     # Furniture
    30,     # Office Supplies
    50      # Technology
])

# Resource constraints
A = np.array([
    [2, 1, 3],      # Labour
    [3, 2, 4]       # Storage
])

b = np.array([
    100,
    150
])

# linprog minimizes, therefore use negative profits
result = linprog(
    -profit_per_unit,
    A_ub=A,
    b_ub=b,
    bounds=[(0, None)] * 3,
    method="highs"
)

print("Optimization Successful:", result.success)

if result.success:

    print(
        "Optimal Units:"
    )

    print(
        "Furniture:",
        result.x[0]
    )

    print(
        "Office Supplies:",
        result.x[1]
    )

    print(
        "Technology:",
        result.x[2]
    )

    print(
        "Maximum Estimated Profit:",
        -result.fun
    )


# ============================================================
# 27. SIMULATION MODELING
# ============================================================

print("\nSIMULATION MODELING")
print("="*50)

np.random.seed(42)

days = 30

daily_demand = np.random.poisson(
    lam=20,
    size=days
)

starting_inventory = 100
inventory = starting_inventory

stockout_days = 0
ending_inventory = []

for demand in daily_demand:

    inventory -= demand

    if inventory < 0:
        stockout_days += 1
        inventory = 0

    ending_inventory.append(
        inventory
    )

print("Total Demand:", daily_demand.sum())
print("Stockout Days:", stockout_days)
print("Ending Inventory:", inventory)

plt.figure(figsize=(10,5))
plt.plot(
    ending_inventory,
    marker="o"
)

plt.title("Inventory Simulation")
plt.xlabel("Day")
plt.ylabel("Inventory")
plt.show()


# ============================================================
# 28. ANOVA
# ============================================================

print("\nANOVA")
print("="*50)

anova_model = ols(
    "Profit ~ C(Category)",
    data=df
).fit()

anova_table = sm.stats.anova_lm(
    anova_model,
    typ=2
)

display(anova_table)

p_anova = anova_table["PR(>F)"][0]

print("ANOVA P-value:", p_anova)

if p_anova < 0.05:
    print(
        "There is a statistically significant difference in profit among categories."
    )
else:
    print(
        "No statistically significant difference was detected among categories."
    )


# ============================================================
# 29. EXPERIMENTAL DESIGN
# ============================================================

print("\nEXPERIMENTAL DESIGN")
print("="*50)

print("""
Experimental Design:

Factor:
Promotion Type

Levels:
1. Email Marketing
2. Social Media Marketing

Response Variable:
Sales

Experimental Units:
Customers

Procedure:
1. Randomly divide customers into two groups.
2. Apply Email Marketing to Group A.
3. Apply Social Media Marketing to Group B.
4. Record sales.
5. Compare average sales.
6. Perform statistical testing.

Purpose:
To identify whether promotion type affects sales.
""")


# ============================================================
# 30. FACTORIAL EXPERIMENT
# ============================================================

print("\nFACTORIAL EXPERIMENT")
print("="*50)

np.random.seed(42)

factorial_data = pd.DataFrame({
    "Promotion": np.repeat(
        ["Email", "Social"],
        40
    ),
    "Discount_Level": np.tile(
        np.repeat(["Low", "High"], 20),
        2
    ),
    "Sales": np.random.normal(
        500,
        100,
        80
    )
})

print("2 × 2 Factorial Experiment:")
display(factorial_data.head())

factorial_model = ols(
    "Sales ~ C(Promotion) * C(Discount_Level)",
    data=factorial_data
).fit()

factorial_anova = sm.stats.anova_lm(
    factorial_model,
    typ=2
)

display(factorial_anova)


# ============================================================
# 31. ADDITIONAL BUSINESS ANALYSIS
# ============================================================

print("\nADDITIONAL BUSINESS ANALYSIS")
print("="*50)


# Average Order Value
average_order_value = (
    total_sales / total_orders
)

print(
    "Average Order Value:",
    round(average_order_value, 2)
)


# Overall Profit Margin
overall_profit_margin = (
    total_profit / total_sales * 100
)

print(
    "Overall Profit Margin:",
    round(overall_profit_margin, 2),
    "%"
)


# Category Profit
category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print("\nCategory Profit:")
display(category_profit)


# Segment Sales
segment_sales = (
    df.groupby("Segment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Segment:")
display(segment_sales)


# Sub-category Sales
subcategory_sales = (
    df.groupby("Sub-Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nSales by Sub-Category:")
display(subcategory_sales.head(10))


# ============================================================
# 32. DISCOUNT ANALYSIS
# ============================================================

print("\nDISCOUNT ANALYSIS")
print("="*50)

df["Discount Band"] = pd.cut(
    df["Discount"],
    bins=[
        -0.01,
        0.10,
        0.20,
        0.30,
        1.00
    ],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)

discount_analysis = (
    df.groupby("Discount Band", observed=False)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Average_Profit=("Profit", "mean")
    )
)

display(discount_analysis)

plt.figure(figsize=(8,5))

discount_analysis["Profit"].plot(
    kind="bar"
)

plt.title("Profit by Discount Band")
plt.xlabel("Discount Band")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()


# ============================================================
# 33. BUSINESS INSIGHTS
# ============================================================

print("\nBUSINESS INSIGHTS AND FINDINGS")
print("="*70)

best_category = category_sales.idxmax()
best_region = region_profit.idxmax()
best_segment = segment_sales.idxmax()
best_subcategory = subcategory_sales.idxmax()

print(
    "1. Highest sales category:",
    best_category
)

print(
    "2. Highest profit region:",
    best_region
)

print(
    "3. Highest sales customer segment:",
    best_segment
)

print(
    "4. Highest sales sub-category:",
    best_subcategory
)

print(
    "5. Total sales:",
    round(total_sales, 2)
)

print(
    "6. Total profit:",
    round(total_profit, 2)
)

print(
    "7. Total orders:",
    total_orders
)

print(
    "8. Total customers:",
    total_customers
)

print(
    "9. Overall profit margin:",
    round(overall_profit_margin, 2),
    "%"
)


# ============================================================
# 34. BUSINESS RECOMMENDATIONS
# ============================================================

print("\nBUSINESS RECOMMENDATIONS")
print("="*70)

print("""
1. Focus on high-performing product categories.

2. Monitor regional performance and improve weaker regions.

3. Track discounts carefully because excessive discounts
   can reduce profitability.

4. Use monthly sales trends for inventory planning.

5. Use forecasting models to support future sales planning.

6. Use customer segmentation to create targeted marketing.

7. Use A/B testing before implementing large marketing campaigns.

8. Use optimization models for better allocation of
   limited resources.

9. Use simulation for inventory and demand planning.

10. Regularly monitor KPIs using a BI dashboard.
""")


# ============================================================
# 35. FINAL PROJECT SUMMARY
# ============================================================

print("\n")
print("="*70)
print("FINAL BUSINESS ANALYTICS PROJECT SUMMARY")
print("="*70)

summary = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Sales",
        "Total Profit",
        "Total Quantity",
        "Total Orders",
        "Total Customers",
        "Average Order Value",
        "Overall Profit Margin",
        "Best Sales Category",
        "Best Profit Region",
        "Best Sales Segment"
    ],
    "Value": [
        len(df),
        round(total_sales, 2),
        round(total_profit, 2),
        total_quantity,
        total_orders,
        total_customers,
        round(average_order_value, 2),
        round(overall_profit_margin, 2),
        best_category,
        best_region,
        best_segment
    ]
})

display(summary)


# ============================================================
# 36. CONCLUSION
# ============================================================

print("\nCONCLUSION")
print("="*70)

print("""
This Business Analytics project analyzed Superstore sales data
using Python and statistical techniques.

The project covered:

• Data Collection
• Data Preprocessing
• Exploratory Data Analysis
• Descriptive Statistics
• Distribution Analysis
• Correlation Analysis
• Regression Analysis
• Business Intelligence
• Probability
• Decision Making Under Uncertainty
• Sampling
• Confidence Intervals
• Bootstrapping
• Hypothesis Testing
• A/B Testing
• Time Series Analysis
• Forecasting
• Forecast Accuracy
• Optimization
• Simulation
• ANOVA
• Experimental Design
• Factorial Experiment

The analysis provides useful business information about sales,
profit, customers, products, regions, discounts and future trends.

The results can support better decision-making, marketing,
inventory planning and resource allocation.
""")

print("="*70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("="*70)
print("KHUSHI PATEL")
```

```python
# ============================================================
# F. DECISION TREE
# ============================================================

from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
import pandas as pd

# Create target variable
# High Profit = Profit greater than or equal to median profit
profit_median = df["Profit"].median()

df["Profit_Level"] = np.where(
    df["Profit"] >= profit_median,
    "High Profit",
    "Low Profit"
)

# Features
X = df[["Sales", "Quantity", "Discount"]]

# Target
y = df["Profit_Level"]

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Create Decision Tree
decision_tree = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

# Train model
decision_tree.fit(X_train, y_train)

# Accuracy
accuracy = decision_tree.score(X_test, y_test)

print("Decision Tree Accuracy:",
      round(accuracy * 100, 2), "%")

# Plot Decision Tree
plt.figure(figsize=(18, 10))

plot_tree(
    decision_tree,
    feature_names=["Sales", "Quantity", "Discount"],
    class_names=["High Profit", "Low Profit"],
    filled=True,
    rounded=True,
    fontsize=10
)

plt.title("Decision Tree for Profit-Level Decision Making")

plt.show()
```
